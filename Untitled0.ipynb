```python
import datetime
import numpy as np
import pandas as pd
import requests
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
import ipywidgets as widgets
from IPython.display import display, clear_output

# --- Base de données enrichie (Massif Central & Sites Populaires France) ---
sites_fr = {
    "Le Villerad (48)": (44.475, 3.450, 950),
    "Truc du Midi (48)": (44.478, 3.455, 1020),
    "Barjac (48) (LOZERE)": (44.505, 3.410, 850),
    "Mende - Brenoux (48)": (44.501, 3.533, 1015),
    "Ispagnac (48)": (44.366, 3.533, 900),
    "St Bonnet de Chirac (48)": (44.518, 3.286, 940),
    "Florac - Rochefort (48)": (44.316, 3.600, 1050),
    "Nasbinals (48)": (44.662, 3.046, 1200),
    "Meyrueis - Brunas (48)": (44.178, 3.428, 980),
    "Puy de Dôme (63)": (45.772, 2.964, 1460),
    "Job - La Forie (63)": (45.633, 3.743, 1050),
    "Valcivière (63)": (45.591, 3.791, 1100),
    "Puy Mary (15) (CANTAL)": (45.108, 2.710, 1780),
    "Le Lioran (15)": (45.084, 2.753, 1250),
    "Le Claux - Enchanet (15)": (45.158, 2.705, 1450),
    "Col de Cabre (15)": (45.122, 2.731, 1520),
    "Dienne (15)": (45.158, 2.848, 1100),
    "Puy de la Tourte (15)": (45.122, 2.695, 1700),
    "Chalinargue (15)": (45.143, 2.930, 1150),
    "Vic-sur-Cère (15)": (44.980, 2.620, 1000),
    "Val de Cère (15)": (45.020, 2.650, 1200),
    "Millau - Puncho d'Agast (12)": (44.116, 3.118, 840),
    "Les Estables - Mézenc (43) (H-LOIRE)": (44.910, 4.190, 1700),
    "Mézenc Sud (43)": (44.900, 4.180, 1650),
    "Alleyras (43)": (44.915, 3.682, 1000),
    "St Jean Lachalm (43)": (44.945, 3.715, 1150),
    "Polignac (43)": (45.071, 3.860, 800),
    "Costaros (43)": (44.883, 3.852, 1100),
    "Dune du Pilat (33)": (44.590, -1.213, 100),
    "Moulis (09) (PYRENEES)": (42.946, 1.091, 1100),
    "Annecy - Montmin (74)": (45.811, 6.242, 1280),
    "St Hilaire du Touvet (38)": (45.308, 5.885, 950),
    "Laragne-Chabre (05)": (44.316, 5.766, 1300),
    "Saint-André-les-Alpes (04)": (43.966, 6.508, 1500),
    "Clécy (14)": (48.916, -0.483, 200),
    "Omaha Beach (14)": (49.367, -0.883, 50)
}

search_box = widgets.Combobox(options=sorted(list(sites_fr.keys())), value="Le Lioran (15)", description='Site:', ensure_option=True, layout={'width': '400px'})

today = datetime.date.today()
tomorrow = today + datetime.timedelta(days=1)
dayaftertomorrow = today + datetime.timedelta(days=2)

day_options = [
    (f"{today.strftime('%A %d %b')} (Aujourd'hui)", 0),
    (f"{tomorrow.strftime('%A %d %b')} (Demain)", 1),
    (f"{dayaftertomorrow.strftime('%A %d %b')} (Après-demain)", 2)
]

day_selector = widgets.Dropdown(options=day_options, value=0, description='Jour:')
out = widgets.Output()

def update_display(site, day_offset):
    if site not in sites_fr: return
    lat, lon, elev = sites_fr[site]
    url = f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&hourly=wind_speed_10m,wind_direction_10m,wind_speed_1000hPa,wind_speed_950hPa,wind_speed_900hPa,wind_speed_850hPa,wind_speed_800hPa,wind_speed_700hPa,wind_speed_600hPa,wind_direction_1000hPa,wind_direction_950hPa,wind_direction_900hPa,wind_direction_850hPa,wind_direction_800hPa,wind_direction_700hPa,wind_direction_600hPa,boundary_layer_height&wind_speed_unit=kmh&timezone=auto"
    with out:
        clear_output(wait=True)
        try:
            resp = requests.get(url).json()["hourly"]
            target_d = today + datetime.timedelta(days=day_offset)
            times = pd.to_datetime(resp["time"])
            # Mise à jour du masque pour aller de 8h à 22h
            mask = (times.date == target_d) & (times.hour >= 8) & (times.hour <= 22)
            idx = np.where(mask)[0]

            if len(idx) == 0:
                print(f"Aucune donnée disponible.")
                return

            levels = [1000, 950, 900, 850, 800, 700, 600]
            alts_ref = np.array([10, 110, 540, 990, 1450, 1950, 3010, 4200])
            target_alts = np.arange(0, 4201, 200)

            fig, ax = plt.subplots(figsize=(14, 8))
            ax.fill_between(range(len(idx)), 0, elev, color='#5D4037', alpha=0.9, zorder=10)

            # --- Tracé du plafond (BLH) ---
            blh = np.array(resp["boundary_layer_height"])[idx] + elev
            ax.plot(range(len(idx)), blh, color='red', linestyle='--', linewidth=2, label='Plafond (BLH)', zorder=25)

            for j, step in enumerate(idx):
                s_vals = np.array([resp["wind_speed_10m"][step]] + [resp[f"wind_speed_{l}hPa"][step] for l in levels])
                d_vals = np.array([resp["wind_direction_10m"][step]] + [resp[f"wind_direction_{l}hPa"][step] for l in levels])

                f_speed = interp1d(alts_ref, s_vals, fill_value="extrapolate", bounds_error=False)
                f_dir = interp1d(alts_ref, d_vals, fill_value="extrapolate", bounds_error=False)

                for alt in target_alts:
                    if alt > elev + 50:
                        s = f_speed(alt); d = f_dir(alt)
                        color = "#90A4AE" if s < 12 else ("#FB8C00" if s < 22 else "#B71C1C")
                        ax.quiver(j, alt, np.cos(np.radians(270-d)), np.sin(np.radians(270-d)), color=color, scale=35, pivot='middle', alpha=0.8, zorder=15)
                        ax.text(j, alt + 100, f"{int(s)}", color='black', fontsize=9, weight='bold', ha='center', va='bottom', zorder=100, bbox=dict(facecolor='white', alpha=0.8, edgecolor='none', pad=1))

            ax.set_title(f"Vent et Plafond pour {site} - {target_d.strftime('%A %d %b')}", fontsize=14, weight='bold')
            ax.set_xticks(range(len(idx))); ax.set_xticklabels(times[mask].strftime('%Hh'))
            ax.set_ylim(0, 4500)
            ax.grid(True, linestyle=':', alpha=0.5)
            ax.legend(loc='upper right')
            plt.show()
            plt.close(fig)
        except Exception as e:
            print(f"Erreur: {e}")

widgets.interactive_output(update_display, {'site': search_box, 'day_offset': day_selector})
display(widgets.VBox([search_box, day_selector, out]))
```

```python
%%writefile app.py
import streamlit as st
import datetime
import numpy as np
import pandas as pd
import requests
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

st.set_page_config(page_title="Météo Parapente", layout="wide")

# --- Base de données ---
sites_fr = {
    "Le Villerad (48)": (44.475, 3.450, 950),
    "Truc du Midi (48)": (44.478, 3.455, 1020),
    "Barjac (48) (LOZERE)": (44.505, 3.410, 850),
    "Mende - Brenoux (48)": (44.501, 3.533, 1015),
    "Ispagnac (48)": (44.366, 3.533, 900),
    "St Bonnet de Chirac (48)": (44.518, 3.286, 940),
    "Florac - Rochefort (48)": (44.316, 3.600, 1050),
    "Nasbinals (48)": (44.662, 3.046, 1200),
    "Meyrueis - Brunas (48)": (44.178, 3.428, 980),
    "Puy de Dôme (63)": (45.772, 2.964, 1460),
    "Puy Mary (15)": (45.108, 2.710, 1780),
    "Le Lioran (15)": (45.084, 2.753, 1250),
    "Dune du Pilat (33)": (44.590, -1.213, 100),
    "Annecy - Montmin (74)": (45.811, 6.242, 1280)
}

st.title("🪂 Windgram Parapente - Application Partageable")

# Sidebar
site_name = st.sidebar.selectbox("Choisir un site", sorted(list(sites_fr.keys())), index=11)
day_offset = st.sidebar.radio("Date", [0, 1, 2], format_func=lambda x: ["Aujourd'hui", "Demain", "Après-demain"][x])

# Calculs
lat, lon, elev = sites_fr[site_name]
today = datetime.date.today()
target_d = today + datetime.timedelta(days=day_offset)

url = f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&hourly=wind_speed_10m,wind_direction_10m,wind_speed_1000hPa,wind_speed_950hPa,wind_speed_900hPa,wind_speed_850hPa,wind_speed_800hPa,wind_speed_700hPa,wind_speed_600hPa,wind_direction_1000hPa,wind_direction_950hPa,wind_direction_900hPa,wind_direction_850hPa,wind_direction_800hPa,wind_direction_700hPa,wind_direction_600hPa,boundary_layer_height&wind_speed_unit=kmh&timezone=auto"

try:
    resp = requests.get(url).json()["hourly"]
    times = pd.to_datetime(resp["time"])
    mask = (times.date == target_d) & (times.hour >= 8) & (times.hour <= 22)
    idx = np.where(mask)[0]

    if len(idx) > 0:
        levels = [1000, 950, 900, 850, 800, 700, 600]
        alts_ref = np.array([10, 110, 540, 990, 1450, 1950, 3010, 4200])
        target_alts = np.arange(0, 4201, 200)

        fig, ax = plt.subplots(figsize=(12, 7))
        ax.fill_between(range(len(idx)), 0, elev, color='#5D4037', alpha=0.9)

        blh = np.array(resp["boundary_layer_height"])[idx] + elev
        ax.plot(range(len(idx)), blh, color='red', linestyle='--', label='Plafond (BLH)')

        for j, step in enumerate(idx):
            s_vals = np.array([resp["wind_speed_10m"][step]] + [resp[f"wind_speed_{l}hPa"][step] for l in levels])
            d_vals = np.array([resp["wind_direction_10m"][step]] + [resp[f"wind_direction_{l}hPa"][step] for l in levels])
            f_speed = interp1d(alts_ref, s_vals, fill_value="extrapolate")
            f_dir = interp1d(alts_ref, d_vals, fill_value="extrapolate")

            for alt in target_alts:
                if alt > elev + 50:
                    s = f_speed(alt); d = f_dir(alt)
                    color = "#90A4AE" if s < 12 else ("#FB8C00" if s < 22 else "#B71C1C")
                    ax.quiver(j, alt, np.cos(np.radians(270-d)), np.sin(np.radians(270-d)), color=color, scale=35, pivot='middle')
                    ax.text(j, alt + 100, f"{int(s)}", color='black', fontsize=8, weight='bold', ha='center', bbox=dict(facecolor='white', alpha=0.7, edgecolor='none', pad=0))

        ax.set_title(f"{site_name} - {target_d.strftime('%A %d %B')}")
        ax.set_xticks(range(len(idx)))
        ax.set_xticklabels(times[mask].strftime('%Hh'))
        ax.set_ylim(0, 4500)
        ax.grid(True, linestyle=':', alpha=0.5)
        st.pyplot(fig)
    else:
        st.warning("Aucune donnée pour cette sélection.")
except Exception as e:
    st.error(f"Erreur API : {e}")
```
